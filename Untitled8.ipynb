```python
#Problem 1: Mobile Recharge Validation System
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        try:
            amount = int(input("Enter recharge amount: "))

            if amount < 50:
                print("Error: Minimum recharge amount is ₹50. Please try again.")
            elif amount not in valid_plans:
                print(f"Error: {amount} is not a valid plan. Choose from {valid_plans}.")
            else:
                print(f"Success: Recharge of ₹{amount} is successful!")
                break # Exit loop on success
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

# Run the function
validate_recharge()
```

```python
#Problem 2: Inventory Reorder Alert System
def check_inventory(inventory):
    for product, stock in inventory.items():
        if stock < 15:
            print(f"Alert: {product} (Stock: {stock}) - Reorder Alert!")
        else:
            print(f"Status: {product} (Stock: {stock}) - Stock OK")

# Sample Data
stock_data = {"Laptops": 20, "Mice": 12, "Keyboards": 45, "Monitors": 8}
check_inventory(stock_data)
```

```python
#Problem 3: Student Result Processing System
def process_student_result(marks_list):
    total = 0
    for mark in marks_list:
        total += mark

    average = total / len(marks_list)

    if average >= 50:
        return f"Average: {average:.2f} - Result: Pass"
    else:
        return f"Average: {average:.2f} - Result: Fail"

# Sample Data
marks = [55, 60, 45, 70, 30]
print(process_student_result(marks))
```

```python
#Problem 4: Cab Fare Estimator with Retry Option
def calculate_fare(km, is_peak):
    base_fare = 50
    rate_per_km = 12
    total = base_fare + (km * rate_per_km)

    if is_peak:
        total *= 1.25 # 25% extra
    return total

while True:
    try:
        distance = float(input("Enter distance in km: "))
        peak = input("Is it peak hour? (yes/no): ").lower() == 'yes'

        fare = calculate_fare(distance, peak)
        print(f"Estimated Fare: ₹{fare:.2f}")

        retry = input("Do you want to calculate another fare? (yes/no): ").lower()
        if retry != 'yes':
            break
    except ValueError:
        print("Invalid input. Please enter a valid distance.")
```

```python
#Problem 5: Employee Attendance Eligibility Checker
def check_attendance_eligibility(attendance_list):
    present_count = 0
    for status in attendance_list:
        if status == "P":
            present_count += 1

    percentage = (present_count / len(attendance_list)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# Sample Data
attendance = ["P", "P", "A", "P", "P", "A", "P"]
print(f"Attendance Status: {check_attendance_eligibility(attendance)}")
```

```python
#Problem 6: Password Strength Checker
def check_password_strength(password):
    if len(password) < 8:
        return "Weak: Password must be at least 8 characters long."

    has_digit = False
    has_special = False
    special_chars = "@#$"

    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak: Must contain at least one digit and one special character (@#$)."

# Test the function
pwd = input("Enter a password to check: ")
print(check_password_strength(pwd))
```
